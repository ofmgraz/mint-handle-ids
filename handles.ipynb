```python
# Define a function to extract and transform values
df4 = pd.read_csv('handles_copy.csv')
def transform_content(cell_content):
    # Match the relevant segment within the content
    match = re.search(r"(A-Gf_[AS][0-9][0-9]*.*)", cell_content)
    if match:
        base_segment = match.group(1).replace(".tif", "")
        if "masters" in cell_content:
            unique_number = f"{base_segment}_M"  # Master
            title = f"{base_segment} (Master)"
        else:
            unique_number = f"{base_segment}_D"  # Derivative
            title = f"{base_segment} (Derivative)"
        return unique_number, title
    return None, None

# Apply the function to extract and fill the columns
df4["eindeutige Nummer des Digitalisats"], df4["Titel des Digitalisats"] = zip(
    *df4["Ressourcenadresse"].apply(lambda x: transform_content(x) if isinstance(x, str) else (None, None))
)
```
